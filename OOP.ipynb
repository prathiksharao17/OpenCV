```python
#built in class functions 
class employee:
    def __init__(self,nm,ag):
        self.name=nm
        self.age=ag

e1=employee('raj',21)
e2=employee('jay',22)

#getattr (get attribute)
print(getattr(e1,'age'))
print(getattr(e2,'age'))

#setattr (set attribute - used to change values)
setattr(e2,'name','jayaraj')
print(e2.__dict__)

#delattr (delete attribute)
delattr(e2,'age')
print(e2.__dict__)

#hasattr (has attribute) 
print(hasattr(e1,'name'))
print(hasattr(e1,'nam'))




```

```python
class Item:
    pay_rate = 0.8 # The pay rate after 20% discount
    all = []
    def __init__(self, name: str, price: float, quantity=0):
        # Run validations to the received arguments
        assert price >= 0, f"Price {price} is not greater than or equal to zero!"
        assert quantity >= 0, f"Quantity {quantity} is not greater or equal to zero!"

        # Assign to self object
        self.name = name
        self.price = price
        self.quantity = quantity

        # Actions to execute
        Item.all.append(self)

    def calculate_total_price(self):
        return self.price * self.quantity

    def apply_discount(self):
        self.price = self.price * self.pay_rate

    def __repr__(self):
        return f"Item('{self.name}', {self.price}, {self.quantity})"

item1 = Item("Phone", 100, 1)
item2 = Item("Laptop", 1000, 3)
item3 = Item("Cable", 10, 5)
item4 = Item("Mouse", 50, 5)
item5 = Item("Keyboard", 75, 5)

print(Item.all)
```

```python
#constructor in inheritance - construction OVERIDING
#Child class can access both child and parent class attributes
#parent class cannot
class Father:
    def __init__(self):
        print("Father Constructor Called")
        self.vehicle='scooter'

class Son(Father):
    pass
s=Son()
print(s.__dict__) #cause son has no constructor

#now if son class also has a  constructor
class Father:
    def __init__(self):
        print("Father Constructor Called")
        self.vehicle='scooter'

class Son(Father):
    def __init__(self):
        self.vehicle='BMW'
s=Son()
print(s.__dict__) #now son has a constructor
#child class constructor has a higher priority than parent class
#if no constructor in child class then parent class attribute will be printed




```

```python
#super()- access parent and child class attributes
class Computer(object):
    def __init__(self):
        self.ram='8gb'
        self.storage='512gb'
        print("Computer class constructor called")
class Mobile(Computer):
    def __init__(self):
        super().__init__()
        self.model='iphone X'
        print("Mobile class constructor called")
apple=Mobile()
print(apple.__dict__)

#if not hardcoded
class Computer(object):
    def __init__(self,ram,storage):
        self.ram=ram
        self.storage=storage
        print("Computer class constructor called")
class Mobile(Computer):
    def __init__(self,ram,storage):
        super().__init__(ram,storage)
        self.model='iphone X'
        print("Mobile class constructor called")
apple=Mobile('8gb','512gb')
print(apple.__dict__)


        
```

```python
#hierarchical inheritance - one parent multiple child
class Person:
    def __init__(self,nm,ag):
        self.name=nm
        self.age=ag
    def display1(self):
        print("THIS IS PERSON DISPLAY")
        
class Employee(Person):
    def __init__(self,nm,ag,sal):
        self.salary=sal
    def display2(self):
        print("THIS IS EMPLOYEE DISPLAY")
class Student(Person):
    def __init__(self,nm,ag,m):
        super().__init__(nm,ag)
        self.marks=m
    def display(self):
        print("THIS IS STUDENT DISPLAY")
s1=Student('jay',21,90)
e1=Employee('raj',23,40000)
p1=Person('raj',23)

s1.display1() #possible
p1.display3() # attribute error

#parent cannot access child attributes but child can access parent attributes




```
